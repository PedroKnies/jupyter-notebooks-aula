### Questão 1: Orçamento de viagem (1,0 ponto)

Escreva um programa que calcule o custo de uma viagem de carro.

O programa deve ler do usuário, usando `input`:

- a distância da viagem, em km (`float`);
- o consumo médio do carro, em km/litro (`float`);
- o preço do litro do combustível, em reais (`float`);
- o valor total dos pedágios, em reais (`float`);

Em seguida o programa deve calcular e exibir:

1. a quantidade de litros necessária (distância dividida pelo consumo);
2. o custo do combustível;
3. o custo total (combustível + pedágios);
4. o custo por pessoa;
5. o tipo (`type`) das variáveis da distância e do número de pessoas.

Todos os valores em reais devem ser exibidos com **2 casas decimais** (use `round` ou formatação de string).

2. calcule `dx` e `dy` como as diferenças **em valor absoluto** entre as coordenadas;
3. calcule a distância entre os pontos com a fórmula abaixo:

$$distancia = \sqrt{dx^2 + dy^2}$$

4. calcule a área do círculo que tem essa distância como raio;
5. exiba `dx`, `dy`, a distância e a área, todos arredondados para 2 casas decimais com `round`.

Exemplo:

```
x1: 1
y1: 2
x2: 4
y2: 6

dx = 3.0
dy = 4.0
```

````python
# Inclua aqui a resposta
import math

#pedi os valores
x1 = float(input("x1: "))
y1 = float(input("y1: "))
x2 = float(input("x2: "))
y2 = float(input("y2: "))

#calculei
dx = abs(x1 - x2)
dy = abs(y1 - y2)
distancia = math.sqrt(pow(dx, 2) + pow(dy, 2))
area = math.pi * pow(distancia, 2)

#mostrei
print("dx =", round(dx, 2))
print("dy =", round(dy, 2))
print("Distância entre os pontos:", round(distancia, 2))
print("a Área do círculo de raio", round(distancia, 2), ":", round(area, 2))
````

**(a)** Escreva uma função chamada `poligono(lados, tamanho)` que desenhe um polígono regular com o número de lados e o comprimento de lado informados. A função deve ter docstring.

**(b)** Escreva uma função chamada `flor(petalas, lados, tamanho)` que **reutilize** `poligono` para desenhar

**(c)** Faça três desenhos: um triângulo, um hexágono e uma flor.

Exemplo:

```

```

````python
# (a) e (b) defina aqui as funções poligono e flor
from jupyturtle import make_turtle, forward, left



def poligono(lados, tamanho):
  """"ele vai pra frente dps vira pra esquerda e pra frente denovo repetindo por cada lado"""
  for i in range(lados):
    forward(tamanho)
    left(360/lados)


def flor(petalas, lados, tamanho):
  for i in range(petalas):
    poligono(lados, tamanho)
    left(360/petalas)
````

**(b)** Escreva uma função chamada `eh_triangulo(a, b, c)` que retorne `True` se três medidas podem formar um
triângulo. A condição é que cada lado seja **positivo** e que cada lado seja **menor** que a soma dos outros dois.

**(c)** Usando `for i in range(...)`, exiba para os anos de **1996 a 2005** se cada um é bissexto ou não.

**(d)** Teste `eh_triangulo` com pelo menos três combinações diferentes, exibindo o resultado de cada teste.

Exemplo:

```

...
2000 é bissexto? True
...
eh_triangulo(3, 4, 5) -> True
eh_triangulo(1, 2, 10) -> False
eh_triangulo(-1, 4, 5) -> False
```

````python
# Inclua aqui a resposta


def eh_bissexto(ano):
    """Retorna True se o ano for bissexto e False caso contrário.

    so é bissexto se for divisivel por 4.
    """
    return (ano % 4 == 0 and ano % 100 != 0) or (ano % 400 == 0)


def eh_triangulo(a, b, c):
    """Retorna True se os lados a, b e c puderem formar um triângulo."""
    lados_positivos = a > 0 and b > 0 and c > 0
    condicao_triangulo = (a < b + c) and (b < a + c) and (c < a + b)
    return lados_positivos and condicao_triangulo



print("--- Teste dos anos bissextos (1996 a 2005) ---")
for ano in range(1996, 2006):  # 2006 para incluir o ano de 2005
    if eh_bissexto(ano):
        print("O ano ", ano, " é bissexto.")
    else:
        print("O ano ", ano," NÃO é bissexto.")




print("Lados (3, 4, 5) formam triângulo?", eh_triangulo(3, 4, 5))

print("Lados (1, 2, 5) formam triângulo?", eh_triangulo(1, 2, 5))

print("Lados (-3, 4, 5) formam triângulo?", eh_triangulo(-3, 4, 5))
````

### Questão 8: Condicionais encadeadas e aninhadas (1,0 ponto)

Escreva uma função chamada `classificar_imc(peso, altura)` que:

1. **valide os dados**: se `peso` **ou** `altura` for menor ou igual a zero, a função deve
   retornar `None`, sem calcular nada;
2. caso os dados sejam válidos, calcule o IMC com a fórmula `imc = peso / altura ** 2`;
3. use uma **condicional encadeada** (`if`/`elif`/`else`) para **retornar** a classificação correspondente:

| IMC | Classificação |
|---|---|
| menor que 18,5 | `"Abaixo do peso"` |
| de 18,5 a 24,9 | `"Peso normal"` |
| de 25,0 a 29,9 | `"Sobrepeso"` |
| 30,0 ou mais | `"Obesidade"` |

Depois, escreva um programa que leia o peso e a altura do usuário, chame a função e:

- se o retorno for `None`, exiba a mensagem `"Dados inválidos: peso e altura devem ser positivos."`;
- caso contrário, exiba o IMC arredondado com 2 casas decimais e a classificação.

A função deve ter **docstring**.

Exemplo:

```
Peso (kg): 72.5
Altura (m): 1.75
IMC: 23.67 -> Peso normal
```

```
Peso (kg): 70
Altura (m): 0
Dados inválidos: peso e altura devem ser positivos.
```

````python
# Inclua aqui a resposta

def classificar_imc(peso, altura):
    """Calcula o indice de Massa Corporal (IMC) e retorna sua classificaçao.

    Retorno:
    menor que 18,5	"Abaixo do peso"
    de 18,5 a 24,9	"Peso normal"
    de 25,0 a 29,9	"Sobrepeso"
    30,0 ou mais	"Obesidade"
    None: Se peso ou altura forem menores ou iguais a zero.
    """

    if peso <= 0 or altura <= 0:
        return None

    # fazer o imc
    imc = peso / (altura**2)


    if imc < 18.5:
        return "Abaixo do peso"
    elif imc <= 24.9:
        return "Peso normal"
    elif imc <= 29.9:
        return "Sobrepeso"
    else:
        return "Obesidade"





peso_input = float(input("Digite o seu peso (em kg): "))
altura_input = float(input("Digite a sua altura (em metros, ex: 1.75): "))


resultado_classificacao = classificar_imc(peso_input, altura_input)

# Verificação do retorno
if resultado_classificacao is None:
    print("Dados inválidos: peso e altura devem ser positivos.")
else:
    # Recalculamos o valor exato apenas para exibir com 2 casas decimais
    imc_calculado = peso_input / (altura_input**2)
    print("seu imc é ", round(imc_calculado, 2), "->", resultado_classificacao)
````
